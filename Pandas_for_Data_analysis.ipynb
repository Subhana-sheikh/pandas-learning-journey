# Create DataFrame  from a Dictionary

```python
data = {
    'Name'   : ['Alice', 'Bob', 'Carol', 'David', 'Eve', 'Frank'],
    'Age'    : [20, 21, 19, 22, 20, 23],
    'Subject': ['Math', 'Science', 'Math', 'English', 'Science', 'English'],
    'Score'  : [88, 73, 55, 40, 95, None],
    'Grade'  : ['A', 'B', 'C', 'F', 'A', None],
    'Passed' : [True, True, True, False, True, None]
}

df = pd.DataFrame(data)
print(df)
```
